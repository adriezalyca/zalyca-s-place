```python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quick_sort(left) + middle + quick_sort(right)

arr = [9, 4, 7, 2, 5, 8, 4, 1]
print("Daftar setelah diurutkan:", quick_sort(arr))
```

```python
def merge_sort(arr):
    if len(arr) > 1:
        mid = len(arr) // 2
        L = arr[:mid]
        R = arr[mid:]

        merge_sort(L)
        merge_sort(R)

        i = j = k = 0

        while i < len(L) and j < len(R):
            if L[i] < R[j]:
                arr[k] = L[i]
                i += 1
            else:
                arr[k] = R[j]
                j += 1
            k += 1

        while i < len(L):
            arr[k] = L[i]
            i += 1
            k += 1

        while j < len(R):
            arr[k] = R[j]
            j += 1
            k += 1

arr = [3, 6, 8, 10, 1, 2, 1]
merge_sort(arr)
print("Daftar setelah diurutkan:", arr)
```

```python
def binary_search(data, target):
    # VALIDASI PRASYARAT: data harus sudah terurut
    assert data == sorted(data), \
        "Error: Binary Search memerlukan data yang sudah terurut!"
    left = 0
    right = len(data) - 1
    while left <= right:
        mid = (left + right) // 2
        if data[mid] == target:
            return mid
        elif data[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1
    # O(log n)
# Data produk olahan agroindustri
produk_olahan = [
    "Yogurt", "Whey Protein", "Susu Pasteurisasi", "Susu Kental Manis",
    "Susu Bubuk", "Mentega", "Kefir", "Keju"
]
# LANGKAH 0 (KRUSIAL): Urutkan data terlebih dahulu
produk_olahan.sort()  # Memastikan prasyarat Binary Search terpenuhi
print("Data setelah diurutkan:", produk_olahan)
target_produk = "Susu Pasteurisasi"
hasil = binary_search(produk_olahan, target_produk)
if hasil != -1:
    print(f"Produk '{target_produk}' ditemukan pada indeks ke-{hasil}")
else:
    print(f"Produk '{target_produk}' tidak ditemukan")
```

```python
# Data hasil panen jagung
data_panen = [
    {"petani": "Pak Slamet", "kadar_air": 14.5},
    {"petani": "Bu Ratna", "kadar_air": 12.8},
    {"petani": "Pak Yono", "kadar_air": 16.2}
]

# Fungsi Bubble Sort berdasarkan kadar air
def bubble_sort(data):
    n = len(data)
    for i in range(n):
        for j in range(0, n-i-1):
            if data[j]["kadar_air"] > data[j+1]["kadar_air"]:
                print(f"Swap: {data[j]['petani']} ({data[j]['kadar_air']}%) "
                      f"<-> {data[j+1]['petani']} ({data[j+1]['kadar_air']}%)")
                # Proses pertukaran data
                data[j], data[j+1] = data[j+1], data[j]
                print("Urutan sementara:")
                for item in data:
                    print(item)
                print()
    return data

# Menampilkan data sebelum pengurutan
print("Data Sebelum Diurutkan:")
for item in data_panen:
    print(item)

print("\nProses Pengurutan:")
hasil = bubble_sort(data_panen)

# Menampilkan data setelah pengurutan
print("\nData Setelah Diurutkan:")
for item in hasil:
    print(item)
```

```python
import time
def insertion_sort(data):
    n = len(data)
    for i in range(1, n):
        current = data[i]
        j = i - 1
        while j >= 0 and data[j] > current:
            data[j + 1] = data[j]
            j -= 1
        data[j + 1] = current
data = [42, 8, 91, 17, 63, 5, 29]
start_time = time.time()
insertion_sort(data)
end_time = time.time()
print("Waktu eksekusi:", end_time - start_time)
print("Data setelah diurutkan:", data)
```
